```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

X_A = pd.get_dummies(dados.drop(columns=['G1', 'G2', 'G3']))
X_B = pd.get_dummies(dados.drop(columns=['G3']))
y = dados['G3']

print(X_A.shape)
print(X_B.shape)
```

```python
sementes = [10, 20, 30, 40, 50]
resultados_A = []
resultados_B = []

for semente in sementes:
    XA_treino, XA_teste, y_treino, y_teste = train_test_split(
        X_A, y, test_size=0.2, random_state=semente
    )

    XB_treino, XB_teste, _, _ = train_test_split(
        X_B, y, test_size=0.2, random_state=semente
    )

    modelo_A = LinearRegression()
    modelo_A.fit(XA_treino, y_treino)
    previsao_A = modelo_A.predict(XA_teste)

    modelo_B = LinearRegression()
    modelo_B.fit(XB_treino, y_treino)
    previsao_B = modelo_B.predict(XB_teste)

    resultados_A.append(mean_absolute_error(y_teste, previsao_A))
    resultados_B.append(mean_absolute_error(y_teste, previsao_B))

print("Modelo A:", resultados_A)
print("Modelo B:", resultados_B)
```

```python
import numpy as np

print("Modelo A")
print("Média:", np.mean(resultados_A))
print("Variância:", np.var(resultados_A, ddof=1))

print("\nModelo B")
print("Média:", np.mean(resultados_B))
print("Variância:", np.var(resultados_B, ddof=1))
```

```python
from scipy.stats import ttest_rel

estatistica, p_valor = ttest_rel(resultados_A, resultados_B)

print("Estatística t:", estatistica)
print("Valor-p:", p_valor)
```

No parecer do grupo, os revisores recomendaram a inclusão da validação cruzada para verificar a consistência dos resultados. Por isso, acrescentei essa etapa ao experimento.

```python
from sklearn.model_selection import KFold, cross_val_score
```

```python
cv = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
mae_cv_A = -cross_val_score(LinearRegression(), X_A, y, cv=cv, scoring='neg_mean_absolute_error')

print(mae_cv_A)
```

```python
mae_cv_B = -cross_val_score(LinearRegression(), X_B, y, cv=cv, scoring='neg_mean_absolute_error')

print(mae_cv_B)
```

```python
import numpy as np

print("Modelo A")
print("Média:", np.mean(mae_cv_A))
print("Variância:", np.var(mae_cv_A, ddof=1))

print("\nModelo B")
print("Média:", np.mean(mae_cv_B))
print("Variância:", np.var(mae_cv_B, ddof=1))
```
